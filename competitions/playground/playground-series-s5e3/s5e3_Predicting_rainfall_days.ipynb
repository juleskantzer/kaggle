<div style="font-family: Segoe UI; line-height: 2; color: #000000; text-align: justify">
  
  <h2 style="color: #162B48">The Dataset</h2>
  <p>The dataset is about weather, specifically precipitation. It comes from the site: <a href="https://www.kaggle.com/competitions/playground-series-s5e3" target="_blank">Kaggle Competition</a>.</p>
  <p>The goal is to determine, based on the characteristics (<mark>pressure, temperature, dew point, humidity</mark>, etc.) of the day, whether it will rain or not.</p>
  <p><mark>The competition starts on March 1, 2025, and ends on March 31, 2025.</mark></p>
  
  <h2 style="color: #162B48">Origin</h2>
  <p>It is always important to know where a dataset comes from, when it was created, and if there is documentation available.</p>
  <p>In our case, we are dealing with a <mark>synthetic dataset</mark>.</p>
  <p><b>What does synthetic data mean?</b></p>
  <p>A synthetic dataset means that the variables are realistic, such as <mark>pressure, temperature, dew point, humidity</mark>, etc., but the data is generated following the distribution of real data.</p>
  <p>For example, imagine a real dataset with <mark>70% women and 30% men</mark> in the “Sex” variable. To create a synthetic dataset, we generate a person’s gender randomly with a probability of <mark>70% for “Female” and 30% for “Male”</mark>. Since the data is generated randomly, we may not get exactly 70% women and 30% men but rather close proportions like 69/31, 65/35, or 68/32. The difference between theoretical values and generated values is called <mark>random error</mark>.</p>
  <p>This process applies not only to the “Sex” variable but also to other variables like “Pressure,” “Temperature,” and “Dew Point,” ensuring that correlations between variables are respected.</p>
  <p><mark>The purpose of a synthetic dataset is to provide realistic data while avoiding exposure of sensitive personal information</mark> such as names, addresses, credit card numbers, etc.</p>
  
  <h2 style="color: #162B48">Files</h2>
  <p>The dataset concerns precipitation in a city or region with <mark>12 variables</mark>: 1 dependent and 11 independent variables. It is divided into two parts:</p>
  <ul>
    <li><b>Training set</b> – Used to train the model.</li>
    <li><b>Test set</b> – Used to evaluate the model.</li>
  </ul>
  <p><mark>The training set contains 2,190 days, while the test set contains 730 days.</mark></p>
  <p>The dataset is relatively small, posing three main issues:</p>
  <ol>
    <li>Machine learning models require a large amount of data. With insufficient data, the model may not converge to the global optimum, negatively affecting performance.</li>
    <li>If the dataset has many dimensions but few data points, data points will be more spaced out. This means the model will have more freedom during training but will generalize poorly to production data, leading to <mark>overfitting</mark>. This problem is known as <b>the curse of dimensionality</b>.</li>
    <li>A small test set makes it easier to manipulate results. In competitions like Kaggle, some participants manually adjust uncertain predictions (probabilities between 40%-60%) to artificially improve scores.</li>
  </ol>
  <p>To address these issues, models that require fewer data points for training, such as <mark>Support Vector Machines (SVMs)</mark>, can be used.</p>
  
  <h2 style="color: #162B48">Data Quality</h2>
  <p><mark>Data quality is essential.</mark> Good quality data reflects real differences, while poor quality data may be distorted by random or systemic errors (biases).</p>
  <p>There are two main types of data:</p>
  <ul>
    <li><b>Objective data</b>: Factual information such as <mark>temperature, weight, height, sex, age, salary, infant mortality rate</mark>, etc.</li>
    <li><b>Subjective data</b>: Abstract or experience-based information like <mark>mental health, depression, stress, GDP, inflation</mark>, etc.</li>
  </ul>
  <p>For example, if a house has 3 bedrooms, that is an <mark>objective</mark> fact. However, if a person is anxious, this is <mark>subjective</mark> and harder to measure.</p>
  <p><b>The “Rainfall” dataset is a good example of using objective data</b> since it contains factual weather-related information.</p>
  
  <h2 style="color: #162B48">Weather Factors</h2>
  <p>Understanding the dataset requires basic knowledge of meteorology. The key weather factors include:</p>
  <ul>
    <li><b>Temperature</b> – Influenced by the Sun, with heat absorbed by the ground and oceans.</li>
    <li><b>Seasons</b> – Result from Earth’s tilt and orbit around the Sun.</li>
    <li><b>Wind</b> – Moving air masses, stronger with faster movement.</li>
    <li><b>Atmospheric Pressure</b> – The weight of air, decreasing with altitude.</li>
    <li><b>Humidity</b> – The amount of water vapor in the air.</li>
    <li><b>Relative Humidity</b> – The ratio of actual water vapor to the maximum possible before condensation.</li>
  </ul>
  <p>These factors are interconnected. For example, <mark>low pressure allows humid air to rise, cool, and condense into clouds and rain</mark>. Conversely, <mark>high pressure compresses air, preventing cloud formation</mark>, leading to sunny weather.</p>
  
  <h2 style="color: #162B48">Conclusion</h2>
  <p><b>All these weather factors are linked</b>. Understanding them helps in interpreting and visualizing the dataset effectively.</p>

```python
# ----- List of accumulated days for each month -----

days_per_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

month_names = ["January", "February", "March", "April", "May", "June",  
               "July", "August", "September", "October", "November", "December"]
```

```python
# ----- Function to create the “month” variable -----

def add_month_column(df):
    df['month'] = df['day'].apply(
        lambda x: month_names[next(i for i, days in enumerate(days_per_month, start=0) 
                                   if x <= sum(days_per_month[:i+1]))]
    )
    return df
```

```python
# ----- Apply the transformation to a DataFrame -----
df_train = add_month_column(df_train)
```

```python
def correlation_matrix_3(df, color_0, color_1, color_2, subtitle_1, subtitle_2, subtitle_3):
    
    # Correlation matrix from -1 to 1 (with 3 colors)
    
    # If we want a correlation matrix from 0 to 1, we need 2 colors
    
    # Import the library: import matplotlib.colors as mcolors
    
    # Highlight text properties
    highlight_textprops = [{"fontsize":12, "color":f'#{color_0}', "fontname": "Cover sans", "fontweight": "bold"},
                           {"fontsize":12, "color":f'#33363F', "fontname": "Cover sans"},
                           {"fontsize":12, "color":f'#{color_0}', "fontname": "Cover sans", "fontweight": "bold"},]


    # Axis labels color
    variable_name_textprops = [{"fontsize":8, "color":f'#33363F', "fontname": "Cover sans", "fontweight": "bold"}]
    
    # Correlation matrix
    correlation_matrix = df.corr(numeric_only=True)

    # Figure
    fig, ax = plt.subplots(figsize=(12, 8), dpi=400)
    

    # Remove the upper half of the matrix
    mask = np.triu(np.ones_like(df.corr(numeric_only=True), dtype=bool))

    
    color1 = mcolors.to_rgba(f'#{color_0}')  # Negative value -1
    color_intermediate = mcolors.to_rgba(f'#{color_1}')  # Intermediate color (Value 0)
    color2 = mcolors.to_rgba(f'#{color_2}')  # Positive value 1

    
    # Create a custom color palette
    n, m = 256, 1
    cmap_custom = mcolors.LinearSegmentedColormap.from_list('custom', [color1, color_intermediate, color2], N=n, gamma=m)

    # Correlation matrix heatmap
    sns.heatmap(correlation_matrix, mask=mask, annot=True, cmap=cmap_custom, fmt=".2f", linewidths=0.2, cbar=False,
               annot_kws={"size": 10})
    
    
    # Horizontal and vertical labels
    xy_label = dict(size=6)
    yticks, ylabels = plt.yticks()
    xticks, xlabels = plt.xticks()
    ax.set_xticklabels(xlabels, rotation=0, **xy_label, **variable_name_textprops[0])
    ax.set_yticklabels(ylabels, **xy_label, **variable_name_textprops[0])
    
    
    # Add a title to the heatmap axis
    ax.set_title('Correlation of Numerical Variables', fontsize=20, fontweight='bold', 
             fontname='Lisboa Sans OSF', color = "#33363F")
    
    
    # Title
    # Ajouter du texte à l'axe avec ax.text()
    ax.text(0.40, 0.845, f"{subtitle_1} {subtitle_2} {subtitle_3}", 
        va='bottom', ha='center', fontsize=12, 
        bbox=dict(facecolor='none', edgecolor='none', boxstyle='round,pad=0.3'), 
        color='black')  # Vous pouvez ajuster les propriétés comme la couleur, la taille de police, etc.
    
```

```python
# ----- Create a transform for a new variable -----

def add_month_column(df):
    df['month'] = df['day'].apply(
        lambda x: month_names[next(i for i, days in enumerate(days_per_month, start=0) 
                                   if x <= sum(days_per_month[:i+1]))]
    )
    return df

month_transformer = FunctionTransformer(add_month_column, validate=False)
```

```python
# ----- Create a transform to delete a variable -----

def remove_day_column(df):
    if 'day' in df.columns:
        df = df.drop(columns=['day'])
    return df

remove_day_transformer = FunctionTransformer(remove_day_column, validate=False)
```

```python
# ----- Create an encoding transformer -----

def apply_get_dummies(X):
    return pd.get_dummies(X, drop_first=True)

encoding = FunctionTransformer(apply_get_dummies, validate=False)
```

```python
# ----- Create a transformer to standardize quantitative variables -----

class SelectiveStandardScaler(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.scaler = StandardScaler()
    
    def fit(self, X, y=None):
        # Sélectionner automatiquement les colonnes numériques
        self.continuous_columns = X.select_dtypes(include=['float64', 'int64']).columns
        self.scaler.fit(X[self.continuous_columns])
        return self
    
    def transform(self, X):
        # Appliquer le scaler seulement sur les colonnes continues
        X_scaled = X.copy()
        X_scaled[self.continuous_columns] = self.scaler.transform(X[self.continuous_columns])
        return X_scaled
```

```python
# ----- Preprocessing Pipeline -----

preprocessing_pipeline = Pipeline([
    
    ('month_feature', month_transformer),   # Add 'Month'
    ('remove_day', remove_day_transformer), # Remove day
    ('encoder', encoding),                  # Encoder
    ('scaler', SelectiveStandardScaler())   # Standardisation
    
])
```

```python
# ----- Testing our pipeline -----

data = {
    'day': np.random.randint(1, 101, size=10),
    'pressure': np.random.uniform(990, 1080, size=10),
    'temparature': np.random.uniform(0, 40, size=10)
}

data = pd.DataFrame(data)

processed_data = preprocessing_pipeline.fit_transform(data)

print("Created data :")
print(data)
print("\n \n")

print("\nData after transformation :")
print(processed_data)
```
